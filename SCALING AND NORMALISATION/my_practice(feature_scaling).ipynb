# Standard Scaler

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
df=pd.read_csv("C:\\Users\\HARSHITA\\Downloads\\Data_sets\\housing.csv")
df.head(2)
```

```python
df=df.drop(columns='Address')
```

```python
x=df.drop(columns=['Price'])
y=df['Price']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
sc=StandardScaler()
```

```python
x_train_sc=sc.fit_transform(x_train)

x_test_sc=sc.transform(x_test)
```

```python
df_x_train=pd.DataFrame(x_train_sc,columns=x_train.columns)
```

```python
df_x_test=pd.DataFrame(x_test_sc,columns=x_test.columns)
```

```python
df_x_test
```

```python
np.round(df_x_train.describe(),2)
```
