# CCA ---> Complete Case Analysis
#### Complete Case analysis simply means dropping missing values.
####                     OR
#### or you can say analysis of those rows whose we had full info.
#### Complete Case Analysis in unbaised when the data in MCAR(Missing Complete At Random)
#### We use Histogram to check weather dropping missing value is affecting our datasets architecture.
#### if past data architecture is overlap to new data architecture ----> Means we can drop the missing data
#### if past data architecture is not overlapped to new data architecture ----> Means we can not drop missing values we will fill the missing values.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("C:\\Users\\HARSHITA\\Downloads\\dsjob - dsjob.csv")
```

```python
df.head(2)
```

```python
df.isnull().mean()
```

```python
cols=[var for var in df.columns if df[var].isnull().mean()<0.05 and df[var].isnull().mean()>0]
cols
```

```python
df[cols].sample(5)
```

```python
df['education_level'].value_counts()
```

```python
len(df[cols].dropna())/len(df)    # we are here getting the percentage that do not have any missing value----> missing_values/df with missing value=non missing values
```

```python
new_df=df[cols].dropna()
df.shape,new_df.shape
```

```python
fig=plt.figure()
ax=fig.add_subplot(111)

# original data

df['experience'].hist(bins=50,ax=ax,density=True,color='Red')

# data after cca the argument alpha make the color transparent ,so we can see the over lay of the 2 distribution

new_df['experience'].hist(bins=50,ax=ax,color='green',density=True,alpha=0.8)
```
