```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("C:\\Users\\HARSHITA\\Downloads\\Data_sets\\world_alcohol.csv")
df.head(4)
```

```python
df.isnull().mean()
```

```python
# applying range
```

```python
cols=[val for val in df.columns if df[val].isnull().mean()<0.50 and df[val].isnull().mean()>0]
cols
```

```python
len(df[cols].dropna())/len(df)
```

```python
new_df=df[cols].dropna()
df.shape,new_df.shape

# I KNOW YOU ARE THINKING THAT CCA DOES NOT REMOVES COUMNS THAN WHY HERE THE COLUMN NUMBER HAS BEEN REDUCES
# GADHI YOU APPLIED THAT DROP ON DF[COLS]----> THAT HAVE ONLY ONE COLUMN .
```

```python
fig=plt.figure()
ax=fig.add_subplot(111)  
df['Display Value'].hist(bins=50,ax=ax,density=True,color='Red')

# BINS=50 ======> DIVIDE DATA RANGE INTO 50 BARS 
#                 WHY 50 ?=====> SMOOTH AND DETAILED DISTRIBUTION

# ax=ax  ======> DRAW HISTOGRAM AT SAME AXIS WITH OUT THIS MATPLOTLIB DRAW AT DIFFERENT AXES

# density=True =====> Area Under Histogram =1 
#            without this----> smaller dataset ---> smaller bars(misleading)

new_df['Display Value'].hist(bins=50,density=True,ax=ax,color='green',alpha=0.8)

# alpha=0.8========> RANGE : 0 (INVISIBLE)
#                    RANGE : 1 (SOLID)

#           WITHOUT THIS----> SECOND HISTOGRAM HIDES THE FIRST

ax.set_title('Distribution of Display Value: Before vs After CCA')
ax.set_xlabel('Display Value')
ax.set_ylabel('Density')
ax.legend()
plt.show()
```
