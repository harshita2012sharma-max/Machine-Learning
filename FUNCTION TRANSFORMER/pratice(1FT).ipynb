```python
import numpy as np
import pandas as pd
```

```python
from sklearn.preprocessing import FunctionTransformer
```

```python
# question-1
salary=np.array([30000,50000,90000,811225])

log_transform=FunctionTransformer(np.log1p)

log_transformed=log_transform.transform(salary)

log_transformed
```

```python
marks=np.array([4,9,16,25,36])
def squ(x):
    return x**2

squ_transform=FunctionTransformer(squ)
transformed=squ_transform.transform(marks)
transformed
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
df = pd.DataFrame({
    'Price': [100, 200, 500, 1000],
    'Quantity': [1, 2, 3, 4]
})

pipe=Pipeline(steps=[
    ('log',FunctionTransformer(np.log1p)),
    ('scaler',StandardScaler())
]
)

price_transformed=pipe.fit_transform(df[['Price']])
print(price_transformed)
```

```python
x=np.array([[3,6,8],[5,6,7],[36,52,12]])

log_transform=FunctionTransformer(np.log1p)

x_transformed=log_transform.transform(x)
x_transformed
```

```python
x=np.array([[12,15],[25,21]])

def squ(x):
    return np.hstack((x,x**2))

squ_transform=FunctionTransformer(squ)

x_transformed=squ_transform.transform(x)
x_transformed
```

```python
x=np.array([[12,24],[13,16]])

def squ(x):
    return np.vstack((x,x**2))

squ_transform=FunctionTransformer(squ)

x_transformed=squ_transform.transform(x)
x_transformed
```

```python
x=np.array([[1,3],[34,np.nan]])

def cleanning(x):
    x[np.isnan(x)]=0
    return x

clean_transform=FunctionTransformer(cleanning)

x_transformed=clean_transform.transform(x)
x_transformed
```
