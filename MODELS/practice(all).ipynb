```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Ridge,Lasso
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score,mean_squared_error
from sklearn.metrics import accuracy_score
```

### Linear Regression

```python
df=pd.read_csv("C:\\Users\\HARSHITA\\Downloads\\Data_sets\\insurance - insurance.csv")
df.head(3)
```

```python
x=df.drop(columns=['charges'])
y=df['charges']
```

#### Label Encoder

```python
le=LabelEncoder()
```

```python
x['sex']=le.fit_transform(x['sex'])
x['smoker']=le.fit_transform(x['smoker'])
x['region']=le.fit_transform(x['region'])
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
lr=LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
y_pred=lr.predict(x_test)
```

```python
r2_score(y_test,y_pred)
```

### Ridge

```python
ridge=Ridge(alpha=0.5)
ridge.fit(x_train,y_train)
```

```python
train_score_ridge=ridge.score(x_train,y_train)
test_score_ridge=ridge.score(x_test,y_test)
print("train_score_ridge",train_score_ridge)
print("test_score_ridge",test_score_ridge)
```

### Lasso

```python
lasso=Lasso(alpha=0.2)
lasso.fit(x_train,y_train)
```

```python
train_score_lasso=lasso.score(x_train,y_train)
test_score_lasso=lasso.score(x_test,y_test)
print("train_score_lasso",train_score_lasso)
print("test_score_lasso",test_score_lasso)
```

### DecisionTreeClassifier

```python
df=pd.read_csv("C:\\Users\\HARSHITA\\Downloads\\Data_sets\\covid_toy - covid_toy.csv")
df.head(2)
df['fever']=df['fever'].fillna(df['fever'].mean())
```

```python
df['gender']=le.fit_transform(df['gender'])
df['cough']=le.fit_transform(df['cough'])
df['city']=le.fit_transform(df['city'])
df['has_covid']=le.fit_transform(df['has_covid'])
```

```python
x=df.drop(columns=['has_covid'])
y=df['has_covid']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
dt=DecisionTreeClassifier()
```

```python
dt.fit(x_train,y_train)
```

```python
from sklearn.metrics import classification_report, roc_auc_score
def evaluate_model(true, predicted):
    score = accuracy_score(true, predicted)
    # report = classification_report(true, predicted)
    roc_auc = roc_auc_score(true, predicted)
    return score, roc_auc
```

```python
models = {
    "Logistic Regression" : LogisticRegression(),
    "Decision Tree Classifier" : DecisionTreeClassifier(),
    "Random Forest Classifier" : RandomForestClassifier()
}

for model_name, model in models.items():
    model.fit(x_train, y_train)

    y_train_pred = model.predict(x_train)
    y_test_pred = model.predict(x_test)

    train_accuracy, train_roc_auc = evaluate_model(y_train, y_train_pred)
    test_accuracy, test_roc_auc = evaluate_model(y_test, y_test_pred)

    print(model_name)

    print("Model Performance for Training Set")
    print("- Accuracy: ", (train_accuracy))
    # print("- Classification Report: {:.4f}".format(train_report))
    print("- ROC AUC Score: ", (train_roc_auc))

    print("-" * 30)

    print("Model Performance for Test Data")
    print("- Accuracy: ", (test_accuracy))
    # print("- Classification Report: {:4f}".format(test_report))
    print("- ROC AUC Score: ", (test_roc_auc))

    print("=" * 30)
```
