```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv("C:\\Users\\HARSHITA\\Downloads\\mall - mall.csv")
df.head(2)
```

```python
df=df.drop(columns=['CustomerID','Genre'])
```

```python
df.head(4)
```

```python
x=df.iloc[:,[0,1]].values
```

```python
from sklearn.cluster import KMeans
```

```python
import matplotlib.pyplot as plt
```

```python
a=[]
for i in range(1,11):
    b=KMeans(n_clusters=i,init='k-means++',random_state=42)
    b.fit(x)
    a.append(b.inertia_)
plt.plot(range(1,11),a)
plt.title('The Elbow Method Graph')
plt.xlabel("Numbers of clusters(k)")
plt.ylabel("wcss_list")
plt.show()
```

```python
b=KMeans(n_clusters=4,init='k-means++',random_state=42)
y_predict=b.fit_predict(x)
```

```python
plt.scatter(x[y_predict==0,0],x[y_predict==0,1],s=50,c='blue',label='Cluster1')
plt.scatter(x[y_predict==1,0],x[y_predict==1,1],s=50,c='green',label='cluster2')
plt.scatter(x[y_predict==2,0],x[y_predict==2,1],s=50,c='yellow',label='cluster3')
plt.scatter(x[y_predict==3,0],x[y_predict==3,1],s=50,c='red',label='cluster4')

plt.scatter(b.cluster_centers_[:,0],b.cluster_centers_[:,1],s=300,c='cyan',label='Centroids')
plt.title("cluster of coustmers")
plt.xlabel("amount income(k$)")
plt.ylabel('spending score(1-100)')
plt.legend()
plt.show()
```
